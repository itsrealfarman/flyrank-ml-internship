## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Question:** Which pages should FlyRank's content team review first for a refresh?

**Unit of analysis:** one page (`content_id`), scored and ranked.

**Output:** a ranked queue of the top-N pages most worth reviewing, each with a reason code explaining why it was flagged.

**Decision this improves:** a content reviewer with limited weekly capacity decides which pages to open first, out of thousands.

**Action someone takes:** the reviewer opens the flagged page and chooses to refresh the content, update metadata, expand it, or leave it as-is — the model only orders the queue, a human makes the final call.

**Cost of a wrong call** (two directions):
- *False positive* (flagged as worth reviewing, but it wasn't): the reviewer spends limited time on a healthy page — the cost is wasted attention, the team's scarcest resource.
- *False negative* (a genuinely declining page never surfaces): the page keeps losing visibility silently, potentially for months, until someone notices by accident.

**Why data/ML helps at all:** with tens of thousands of pages and growing, no human can manually review everything every week. The model doesn't replace the reviewer's judgment — it triages, so limited human attention goes to the pages most likely to be worth it first.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*


```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: how many pages are currently flagged as declining
declining = (df["trend_direction"].str.lower() == "down").sum()
total = len(df)
print(f"Declining pages: {declining} of {total} ({declining/total*100:.1f}%)")

declining_df = df[df["trend_direction"].str.lower() == "down"]

# Number 2: total exposure sitting on declining pages
total_impressions_at_risk = declining_df["impressions_90d"].sum()
print(f"Total impressions_90d sitting on declining pages: {total_impressions_at_risk:,.0f}")

# Number 3: how few declining pages an obvious hand-written filter alone would catch
stale_and_declining = declining_df[
    (declining_df["days_since_last_update"] >= 180) &
    (declining_df["impressions_90d"] >= 500)
]
print(f"Declining AND stale+visible (what an obvious rule alone catches): "
      f"{len(stale_and_declining)} of {len(declining_df)} declining pages")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

**What I can claim:** on this starter slice, a trained model produced a better-ordered review queue than a simple hand-written rule, measured by Precision@50 with client-holdout validation (whole clients held out of training). This is decision-support — it helps a reviewer prioritize a limited queue, using observed, directional evidence.

**What I cannot claim:**
- That refreshing a flagged page will *cause* traffic to recover — proving that needs a controlled experiment or causal design, which this data alone cannot give me.
- That this result generalizes to the full ~79M-row warehouse — it has to be re-earned there with proper validation, not assumed from a 30,000-row starter slice.
- That any signal here reflects a real Google ranking factor — I'm working with observable search/content signals, not Google's algorithm.
- That "declining" always means a real, sustained decline — before treating a drop as real, I need to rule out consolidation (a sibling page absorbing demand), seasonality, and simple low-volume noise.

I'll keep language observed/directional throughout ("the data suggests", "this is associated with") rather than causal ("this proves", "this causes").
